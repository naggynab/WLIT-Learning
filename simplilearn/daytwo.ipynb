Short Git list:

```powershell
git status
```

See what changed.

```powershell
git add .
```

Stage all new, modified, and deleted files in the current folder.

```powershell
git add filename
```

Stage one file.

```powershell
git add -u
```

Stage changes and deletions to files Git already knows about.

It does **not** stage new untracked files.

Example:

```powershell
git add -u -- daytwo.txt
```

Stages the deletion or modification of `daytwo.txt`.

```powershell
git commit -m "message"
```

Save staged changes locally.

```powershell
git push origin main
```

Upload commits to GitHub.

Memory:

```text
status → add → commit → push
see → prepare → save → upload
```
